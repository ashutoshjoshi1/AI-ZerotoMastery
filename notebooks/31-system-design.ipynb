### 7. Operations

*Latency budget, cost per request and per month, monitoring, model upgrades.* Use the estimator below.

```python
def estimate(requests_per_day, in_tokens, out_tokens, price_in_per_m, price_out_per_m,
             ttft_s=0.6, tokens_per_s=60):
    """Back-of-envelope monthly cost and typical latency for one model call per request."""
    cost = (in_tokens / 1e6 * price_in_per_m + out_tokens / 1e6 * price_out_per_m) * requests_per_day * 30
    latency = ttft_s + out_tokens / tokens_per_s
    return {"monthly_cost": round(cost, 2), "latency_s": round(latency, 2)}

# Plug in your provider's current prices (per million tokens) and your own measurements:
print(estimate(2000, in_tokens=4000, out_tokens=350, price_in_per_m=3.0, price_out_per_m=15.0))
```
