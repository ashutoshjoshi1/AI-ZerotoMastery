## How the field fits together

The vocabulary is nested. Artificial intelligence contains machine learning, which contains deep learning, which contains large language models. A chess engine that searches moves with hand-written rules is AI, but it does not learn. Machine learning means the behaviour comes from data: you show a spam filter labelled emails and it works out the rules itself.

The difference is easiest to feel in code, so the first exercise asks you to build both a rule-based and a learned spam filter.

### Hands-on: rules versus learning

Using only the Python standard library:

1. Write twelve short labelled messages (half spam, half not) as a `train` list of `(text, label)` pairs, and eight more as a `test` list. Make some test spam use words that never appear in your rule's keyword list.
2. Write `rule_filter(text)`: spam if the message contains any of a hand-picked set of keywords such as `free`, `win`, `prize`.
3. Write `learned_filter(text)`: count, in the training messages, how often each word appears in spam and in non-spam. Score a message as the sum over its words of `log((spam_count + 1) / (ham_count + 1))` and call it spam if the score is positive.
4. Measure both on the test list. You are done when the learned filter beats the rule, and you can explain why it catches a spam message that contains none of your keywords.

## Set up your workbench

### Python and environments

**Option: uv (recommended)**

Use Python 3.12. Anything from 3.11 up works, but the newest Python release often lags behind in library support, so 3.12 is the safe choice. The easiest modern way to manage Python versions and packages is [uv](https://docs.astral.sh/uv/), a fast replacement for `pip` and `venv` that also installs Python itself. The classic `venv` plus `pip` route works just as well; both are shown.

Run in a terminal:

```bash
# install uv (macOS / Linux)
curl -LsSf https://astral.sh/uv/install.sh | sh
# Windows (PowerShell): powershell -ExecutionPolicy ByPass -c "irm https://astral.sh/uv/install.ps1 | iex"

mkdir ai-engineering-log && cd ai-engineering-log
uv init --python 3.12          # creates pyproject.toml and a .venv on first sync
uv add numpy pandas matplotlib scikit-learn jupyterlab python-dotenv
uv run jupyter lab             # runs inside the project's environment
```

**Option: venv + pip**

Run in a terminal:

```bash

source .venv/bin/activate      # Windows: .venv\Scripts\activate
```

### API keys, safely

In part VI you will call hosted models. Their API keys are secrets that cost you money if leaked. **Never paste a key into code or a notebook.** Put it in a `.env` file that Git ignores, and load it at runtime.

In a file called `.env`:

```bash
ANTHROPIC_API_KEY=sk-ant-...
OPENAI_API_KEY=sk-...
```

````python
from dotenv import load_dotenv
import os

load_dotenv()                      # reads .env into environment variables
key = os.environ.get("ANTHROPIC_API_KEY")
print("key loaded" if key else "no key found")   # never print the key itself
````

> **If a key leaks**
>
> Revoke it in the provider's console immediately, then create a new one. Bots scan public GitHub repositories for keys within minutes of a push. Rewriting Git history afterwards is not enough, because the key has already been copied.
