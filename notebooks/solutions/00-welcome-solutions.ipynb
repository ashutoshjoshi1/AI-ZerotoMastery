## Set up your workbench

### Python and environments

**Option: uv (recommended)**

Use Python 3.11 or newer. The easiest modern way to manage Python versions and packages is [uv](https://docs.astral.sh/uv/), a fast replacement for `pip` and `venv`. The classic `venv` plus `pip` route works just as well; both are shown.

Run in a terminal:

```bash
# install uv (macOS / Linux); see the uv docs for Windows
curl -LsSf https://astral.sh/uv/install.sh | sh

mkdir ai-engineering-log && cd ai-engineering-log
uv init --python 3.12          # creates pyproject.toml and a .venv on first sync
uv add numpy pandas matplotlib scikit-learn jupyterlab python-dotenv
uv run jupyter lab             # runs inside the project's environment
```

**Option: venv + pip**

Run in a terminal:

```bash
mkdir ai-engineering-log && cd ai-engineering-log
python3 -m venv .venv
source .venv/bin/activate      # Windows: .venv\Scripts\activate
pip install --upgrade pip
pip install numpy pandas matplotlib scikit-learn jupyterlab python-dotenv
jupyter lab
```

### API keys, safely

In part VI you will call hosted models. Their API keys are secrets that cost you money if leaked. **Never paste a key into code or a notebook.** Put it in a `.env` file that Git ignores, and load it at runtime.

Run in a terminal:

```bash
ANTHROPIC_API_KEY=sk-ant-...
OPENAI_API_KEY=sk-...
```

````python
from dotenv import load_dotenv
import os

load_dotenv()                      # reads .env into environment variables
key = os.environ.get("ANTHROPIC_API_KEY")
print("key loaded" if key else "no key found")   # never print the key itself
````

> **If a key leaks**
>
> Revoke it in the provider's console immediately, then create a new one. Bots scan public GitHub repositories for keys within minutes of a push. Rewriting Git history afterwards is not enough, because the key has already been copied.

### Hands-on: prove your workbench works

Create `src/check_env.py` that prints your Python version and the versions of NumPy, pandas, scikit-learn and Matplotlib, then reports whether an API key is present (without printing it). Run it with `uv run python src/check_env.py` or `python src/check_env.py`. Commit the file and write your first log entry.

#### Solution: a solution

````python
%%writefile check_env.py
import os
import sys
from importlib.metadata import version, PackageNotFoundError

from dotenv import load_dotenv

load_dotenv()
print(f"Python {sys.version.split()[0]}")
for pkg in ["numpy", "pandas", "scikit-learn", "matplotlib", "jupyterlab"]:
    try:
        print(f"{pkg:<14}{version(pkg)}")
    except PackageNotFoundError:
        print(f"{pkg:<14}MISSING: install it with uv add {pkg}")

for var in ["ANTHROPIC_API_KEY", "OPENAI_API_KEY"]:
    status = "present" if os.environ.get(var) else "not set (fine until part VI)"
    print(f"{var:<20}{status}")
````

````python
%run check_env.py
````

`importlib.metadata.version` reads installed package metadata, which works even for packages whose import name differs from their install name (scikit-learn installs as `sklearn`).
